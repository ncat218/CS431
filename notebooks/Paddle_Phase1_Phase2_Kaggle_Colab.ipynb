```python
CUDA_WHEEL = 'cu126'
VENV = BASE / 'cs431-paddle'
subprocess.run([sys.executable, str(REPO/'tools/setup_paddle.py'), '--venv', str(VENV), '--cuda', CUDA_WHEEL], cwd=REPO, check=True)
PYTHON = str(VENV/'bin/python')
subprocess.run([PYTHON, str(REPO/'tools/prepare_paddle_model.py')], cwd=REPO, check=True)
GPUS = '0,1' if Path('/kaggle/working').exists() else '0'
print('GPU IDs:', GPUS, '(T4x2: 0,1; one A100: 0)')
```

## Phase 1 — khai báo dữ liệu thật
Cần đủ img_1 … img_500 và gt_img_1.txt … gt_img_500.txt. Không dùng train split.
Liệt kê thư mục trước, rồi điền đường dẫn và nguồn ICDAR2015 đã attach. Không suy đoán đường dẫn.


```python
INPUT_ROOT = Path('/kaggle/input') if Path('/kaggle/input').exists() else Path('/content')
print('Available input directories:')
for item in sorted(INPUT_ROOT.iterdir()):
    if item.is_dir(): print(item)
ICDAR_IMAGES = ''  # Thư mục chứa 500 test images (có thể tìm đệ quy bên trong)
ICDAR_GT = ''      # Thư mục chứa 500 localization/transcription GT .txt
DATASET_SOURCE = ''  # URL/phiên bản/nguồn dữ liệu thực tế, để reviewer đối chiếu
if not ICDAR_IMAGES or not ICDAR_GT or not DATASET_SOURCE:
    raise ValueError('Điền ba trường ICDAR_IMAGES, ICDAR_GT và DATASET_SOURCE trước khi chạy Phase 1.')
subprocess.run([PYTHON, str(REPO/'run.py'), 'paddle','phase1',
    '--images',ICDAR_IMAGES,'--gt',ICDAR_GT,'--dataset-source',DATASET_SOURCE,'--gpus',GPUS], cwd=REPO, check=True)
```

## Người thực hiện ghi nhận đánh giá Phase 1

## Phase 2 — local roots và cùng 9.898 cặp
Root phải ghép được với relative paths trong manifest. Không sửa ID/manifest để chữa đường dẫn.
Chỉ hai root ảnh cần cho inference; TextOCR GT subset đã đóng trong frozen manifest.
Giữ đủ ba dataset/metadata nguồn để audit khi có vấn đề. Cell liệt kê ví dụ relative path để chọn root thật.


```python
manifest = REPO/'outputs/shared/shared_manifest_v1.jsonl'
with manifest.open() as f: example = json.loads(next(f))
print('Clean relative path:',example['clean_relpath'])
print('Degraded relative path:',example['degraded_relpath'])
CLEAN_ROOT = ''
DEGRADED_ROOT = ''
if not CLEAN_ROOT or not DEGRADED_ROOT:
    raise ValueError('Điền CLEAN_ROOT và DEGRADED_ROOT theo file thật đã attach.')
locations = REPO/'outputs/dataset_locations.json'
locations.parent.mkdir(parents=True,exist_ok=True)
locations.write_text(json.dumps(dict(clean_root=CLEAN_ROOT,degraded_root=DEGRADED_ROOT)))
subprocess.run([PYTHON,str(REPO/'run.py'),'paddle','phase2','--locations',str(locations),'--gpus',GPUS],cwd=REPO,check=True)
```
